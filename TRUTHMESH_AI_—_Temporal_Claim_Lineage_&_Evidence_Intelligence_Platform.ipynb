```python
# ================================================================
#                    TRUTHMESH AI v2.0
# ================================================================
# Temporal Claim Lineage & Evidence Intelligence Platform
#
# Research Question:
#
# "When a scientific claim travels across sources and time,
#  how do its meaning, certainty, context and evidence change?"
#
# ================================================================
#
# CORE MODULES
# ---------------------------------------------------------------
# 1. Claim ingestion
# 2. NLP preprocessing
# 3. Entity extraction
# 4. Modality / certainty detection
# 5. Qualifier extraction
# 6. Subject-Predicate-Object extraction
# 7. Claim DNA
# 8. Semantic representation
# 9. Claim similarity
# 10. Claim family detection
# 11. Temporal lineage
# 12. Claim drift
# 13. Context retention
# 14. Information decay
# 15. Evidence-claim divergence
# 16. Source dependency
# 17. Contradiction detection
# 18. Claim mutation analysis
# 19. Knowledge graph
# 20. Research Copilot
# 21. Claim Search
# 22. Claim Time Machine
# 23. Interactive visual analytics
# 24. CSV / JSON export
#
# ================================================================


# ================================================================
# 0. INSTALLATION
# ================================================================

import sys
import subprocess
import importlib
import os
import re
import json
import math
import warnings
from collections import Counter, defaultdict

warnings.filterwarnings("ignore")


def ensure_package(package_name, import_name=None):

    import_name = import_name or package_name

    try:
        importlib.import_module(import_name)

    except ImportError:

        print(f"Installing {package_name}...")

        subprocess.check_call([
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            package_name
        ])


REQUIRED_PACKAGES = [
    ("pandas", "pandas"),
    ("numpy", "numpy"),
    ("scikit-learn", "sklearn"),
    ("networkx", "networkx"),
    ("plotly", "plotly")
]


for package_name, import_name in REQUIRED_PACKAGES:
    ensure_package(package_name, import_name)


# ================================================================
# 1. IMPORTS
# ================================================================

import pandas as pd
import numpy as np
import networkx as nx

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

import plotly.express as px
import plotly.graph_objects as go

from IPython.display import display, Markdown


# ================================================================
# 2. CONFIGURATION
# ================================================================

PROJECT_NAME = "TRUTHMESH AI"

VERSION = "2.0.0"

OUTPUT_DIR = "truthmesh_v2_outputs"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

RANDOM_STATE = 42

SIMILARITY_THRESHOLD = 0.32

HIGH_SIMILARITY_THRESHOLD = 0.55


# ================================================================
# 3. HEADER
# ================================================================

print("\n")
print("=" * 90)
print(f"                    {PROJECT_NAME}")
print(f"          Temporal Claim Lineage & Evidence Intelligence")
print(f"                         VERSION {VERSION}")
print("=" * 90)


# ================================================================
# 4. DEMONSTRATION DATASET
# ================================================================
#
# This dataset intentionally contains claims whose strength and
# wording change over time.
#
# Production version:
# Replace this section with PubMed / Crossref / arXiv / CSV / API
# ingestion.
#
# ================================================================

records = [

    # ------------------------------------------------------------
    # FAMILY A
    # ------------------------------------------------------------

    {
        "claim_id": "C001",
        "date": "2021-01-10",
        "source": "Research Paper A",
        "source_type": "Peer Reviewed",
        "author": "Smith et al.",
        "claim": "Drug X may reduce inflammation in patients with chronic disease.",
        "evidence_strength": 0.72,
        "sample_size": 120
    },

    {
        "claim_id": "C002",
        "date": "2021-06-15",
        "source": "Research Paper B",
        "source_type": "Peer Reviewed",
        "author": "Jones et al.",
        "claim": "Drug X appears to reduce inflammation in patients with chronic disease.",
        "evidence_strength": 0.68,
        "sample_size": 95
    },

    {
        "claim_id": "C003",
        "date": "2022-02-20",
        "source": "Medical Review Website",
        "source_type": "Review",
        "author": "Medical Review Team",
        "claim": "Drug X reduces inflammation in patients with chronic disease.",
        "evidence_strength": 0.55,
        "sample_size": 95
    },

    {
        "claim_id": "C004",
        "date": "2023-05-01",
        "source": "Health News Article",
        "source_type": "News",
        "author": "Health Desk",
        "claim": "Drug X is proven to reduce inflammation in patients with chronic disease.",
        "evidence_strength": 0.42,
        "sample_size": 95
    },


    # ------------------------------------------------------------
    # FAMILY B
    # ------------------------------------------------------------

    {
        "claim_id": "C005",
        "date": "2019-03-12",
        "source": "Genomics Study A",
        "source_type": "Peer Reviewed",
        "author": "Lee et al.",
        "claim": "Gene X may be associated with disease Y.",
        "evidence_strength": 0.81,
        "sample_size": 500
    },

    {
        "claim_id": "C006",
        "date": "2020-07-19",
        "source": "Genomics Study B",
        "source_type": "Peer Reviewed",
        "author": "Patel et al.",
        "claim": "Gene X is associated with disease Y.",
        "evidence_strength": 0.79,
        "sample_size": 460
    },

    {
        "claim_id": "C007",
        "date": "2021-11-03",
        "source": "Biomedical Blog",
        "source_type": "Blog",
        "author": "Science Writer",
        "claim": "Gene X contributes to disease Y.",
        "evidence_strength": 0.51,
        "sample_size": 200
    },

    {
        "claim_id": "C008",
        "date": "2023-01-25",
        "source": "Popular Science Article",
        "source_type": "Media",
        "author": "Science Desk",
        "claim": "Gene X causes disease Y.",
        "evidence_strength": 0.35,
        "sample_size": 200
    },


    # ------------------------------------------------------------
    # FAMILY C
    # ------------------------------------------------------------

    {
        "claim_id": "C009",
        "date": "2022-02-11",
        "source": "Developer Study A",
        "source_type": "Peer Reviewed",
        "author": "Research Group A",
        "claim": "AI coding assistants may improve developer productivity.",
        "evidence_strength": 0.74,
        "sample_size": 300
    },

    {
        "claim_id": "C010",
        "date": "2022-12-09",
        "source": "Developer Study B",
        "source_type": "Peer Reviewed",
        "author": "Research Group B",
        "claim": "AI coding assistants can improve developer productivity.",
        "evidence_strength": 0.71,
        "sample_size": 250
    },

    {
        "claim_id": "C011",
        "date": "2023-08-14",
        "source": "Technology Blog",
        "source_type": "Blog",
        "author": "Tech Writer",
        "claim": "AI coding assistants improve developer productivity.",
        "evidence_strength": 0.48,
        "sample_size": 100
    },

    {
        "claim_id": "C012",
        "date": "2024-03-22",
        "source": "Technology News",
        "source_type": "News",
        "author": "Technology Desk",
        "claim": "AI coding assistants have been proven to dramatically improve developer productivity.",
        "evidence_strength": 0.39,
        "sample_size": 100
    },


    # ------------------------------------------------------------
    # FAMILY D
    # ------------------------------------------------------------

    {
        "claim_id": "C013",
        "date": "2020-04-10",
        "source": "Clinical Trial Y",
        "source_type": "Clinical Trial",
        "author": "Clinical Team",
        "claim": "Drug Y did not significantly reduce inflammation in the trial population.",
        "evidence_strength": 0.91,
        "sample_size": 850
    },

    {
        "claim_id": "C014",
        "date": "2021-09-18",
        "source": "Animal Research Y",
        "source_type": "Animal Study",
        "author": "Animal Research Team",
        "claim": "Drug Y reduced inflammation in an animal model.",
        "evidence_strength": 0.77,
        "sample_size": 80
    },


    # ------------------------------------------------------------
    # FAMILY E
    # ------------------------------------------------------------

    {
        "claim_id": "C015",
        "date": "2018-01-12",
        "source": "Pilot Study",
        "source_type": "Preliminary",
        "author": "Pilot Research Team",
        "claim": "Preliminary observations suggest that Gene A may influence disease X.",
        "evidence_strength": 0.32,
        "sample_size": 32
    },

    {
        "claim_id": "C016",
        "date": "2022-06-30",
        "source": "Science Blog",
        "source_type": "Blog",
        "author": "Science Blogger",
        "claim": "Gene A influences disease X.",
        "evidence_strength": 0.25,
        "sample_size": 32
    }
]


claims = pd.DataFrame(records)

claims["date"] = pd.to_datetime(
    claims["date"]
)

claims = claims.sort_values(
    "date"
).reset_index(drop=True)


print("\nDATASET LOADED")
print("-" * 90)
print("Claims:", len(claims))
print("Sources:", claims["source"].nunique())
print(
    "Date range:",
    claims["date"].min().date(),
    "→",
    claims["date"].max().date()
)


# ================================================================
# 5. TEXT NORMALIZATION
# ================================================================

def normalize_text(text):

    text = str(text).lower()

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    text = re.sub(
        r"[^\w\s\-]",
        "",
        text
    )

    return text.strip()


claims["normalized_claim"] = (
    claims["claim"]
    .apply(normalize_text)
)


# ================================================================
# 6. ENTITY EXTRACTION
# ================================================================

def extract_entities(text):

    text = str(text)

    entities = []

    patterns = [

        r"\bDrug\s+[A-Z][A-Za-z0-9\-]*",

        r"\bGene\s+[A-Z][A-Za-z0-9\-]*",

        r"\bdisease\s+[A-Za-z0-9\-]+",

    ]

    for pattern in patterns:

        matches = re.findall(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        entities.extend(matches)

    if re.search(
        r"\bAI\b",
        text
    ):

        entities.append("AI")

    if "coding assistants" in text.lower():

        entities.append(
            "coding assistants"
        )

    if "inflammation" in text.lower():

        entities.append(
            "inflammation"
        )

    if "productivity" in text.lower():

        entities.append(
            "developer productivity"
        )

    # Deduplicate while preserving order
    return list(
        dict.fromkeys(
            entities
        )
    )


claims["entities"] = (
    claims["claim"]
    .apply(extract_entities)
)


# ================================================================
# 7. MODALITY / CERTAINTY
# ================================================================

MODALITY_LEVELS = {

    "may": 0.25,

    "preliminary": 0.30,

    "appears": 0.40,

    "can": 0.55,

    "associated": 0.55,

    "influences": 0.65,

    "contributes": 0.70,

    "reduces": 0.75,

    "improves": 0.75,

    "causes": 0.90,

    "proven": 0.95
}


def detect_modality(text):

    t = text.lower()

    if "proven" in t:
        return "proven"

    if "causes" in t:
        return "causal"

    if "contributes" in t:
        return "contribution"

    if "influences" in t:
        return "influence"

    if "associated" in t:
        return "association"

    if "appears" in t:
        return "appears"

    if "preliminary" in t:
        return "preliminary"

    if "may" in t:
        return "uncertain"

    if "can" in t:
        return "capability"

    if "did not" in t:
        return "negative"

    return "assertion"


claims["modality"] = (
    claims["claim"]
    .apply(detect_modality)
)


def modality_score(modality):

    mapping = {

        "uncertain": 0.25,

        "preliminary": 0.30,

        "appears": 0.40,

        "capability": 0.55,

        "association": 0.55,

        "influence": 0.65,

        "contribution": 0.70,

        "assertion": 0.70,

        "negative": 0.75,

        "causal": 0.90,

        "proven": 0.95
    }

    return mapping.get(
        modality,
        0.50
    )


claims["modality_score"] = (
    claims["modality"]
    .apply(modality_score)
)


# ================================================================
# 8. QUALIFIER / CONTEXT EXTRACTION
# ================================================================

QUALIFIERS = [

    "may",

    "might",

    "appears",

    "preliminary",

    "suggest",

    "suggests",

    "possibly",

    "significantly",

    "not significantly",

    "dramatically",

    "proven",

    "in patients",

    "animal model",

    "trial population",

    "chronic disease"
]


def extract_qualifiers(text):

    t = text.lower()

    found = []

    for q in QUALIFIERS:

        if q in t:

            found.append(q)

    return list(
        dict.fromkeys(found)
    )


claims["qualifiers"] = (
    claims["claim"]
    .apply(extract_qualifiers)
)


# ================================================================
# 9. SUBJECT
# ================================================================

def extract_subject(text):

    t = text.lower()

    entities = extract_entities(
        text
    )

    if "ai coding assistants" in t:

        return "AI coding assistants"

    if entities:

        # Prefer biological entity
        for entity in entities:

            if (
                entity.lower().startswith("gene")
                or
                entity.lower().startswith("drug")
            ):

                return entity

    return "Unknown"


claims["subject"] = (
    claims["claim"]
    .apply(extract_subject)
)


# ================================================================
# 10. OBJECT
# ================================================================

def extract_object(text):

    t = text.lower()

    if "inflammation" in t:

        return "inflammation"

    if "developer productivity" in t:

        return "developer productivity"

    entities = extract_entities(
        text
    )

    if len(entities) >= 2:

        return entities[-1]

    return "Unknown"


claims["object"] = (
    claims["claim"]
    .apply(extract_object)
)


# ================================================================
# 11. PREDICATE
# ================================================================

def extract_predicate(text):

    t = text.lower()

    if "causes" in t:
        return "causes"

    if "contributes" in t:
        return "contributes to"

    if "influences" in t:
        return "influences"

    if "associated" in t:
        return "associated with"

    if "reduce" in t:
        return "reduces"

    if "improve" in t:
        return "improves"

    if "did not" in t:
        return "does not significantly reduce"

    return "relates to"


claims["predicate"] = (
    claims["claim"]
    .apply(extract_predicate)
)


# ================================================================
# 12. CLAIM STRENGTH
# ================================================================

def calculate_claim_strength(row):

    evidence = float(
        row["evidence_strength"]
    )

    modality = float(
        row["modality_score"]
    )

    # Sample size normalization
    sample = float(
        row["sample_size"]
    )

    sample_score = min(
        math.log1p(sample)
        /
        math.log1p(1000),
        1.0
    )

    strength = (

        0.45 * modality

        +

        0.40 * evidence

        +

        0.15 * sample_score

    )

    return round(
        strength,
        4
    )


claims["claim_strength"] = (
    claims.apply(
        calculate_claim_strength,
        axis=1
    )
)


# ================================================================
# 13. CLAIM FAMILY DETECTION
# ================================================================
#
# For this demonstration dataset we use semantic anchor entities.
#
# A production version would use:
#
# Sentence Transformers
# SciBERT / PubMedBERT
# Embedding clustering
# Entity linking
#
# ================================================================

def determine_family(row):

    text = row["claim"].lower()

    if "drug x" in text:

        return "Drug X – Inflammation"

    if "gene x" in text:

        return "Gene X – Disease Y"

    if "ai coding assistants" in text:

        return "AI – Developer Productivity"

    if "drug y" in text:

        return "Drug Y – Inflammation"

    if "gene a" in text:

        return "Gene A – Disease X"

    return "Unclassified"


claims["family"] = (
    claims.apply(
        determine_family,
        axis=1
    )
)


# ================================================================
# 14. CLAIM DNA
# ================================================================

def create_claim_dna(row):

    return {

        "claim_id":
            row["claim_id"],

        "subject":
            row["subject"],

        "predicate":
            row["predicate"],

        "object":
            row["object"],

        "modality":
            row["modality"],

        "qualifiers":
            row["qualifiers"],

        "entities":
            row["entities"],

        "claim_strength":
            float(row["claim_strength"]),

        "evidence_strength":
            float(row["evidence_strength"]),

        "source":
            row["source"],

        "source_type":
            row["source_type"]
    }


claims["claim_dna"] = (
    claims.apply(
        create_claim_dna,
        axis=1
    )
)


# ================================================================
# 15. SEMANTIC VECTORS
# ================================================================

vectorizer = TfidfVectorizer(

    stop_words="english",

    ngram_range=(1, 2),

    min_df=1,

    sublinear_tf=True
)


claim_matrix = vectorizer.fit_transform(
    claims["normalized_claim"]
)


similarity_matrix = cosine_similarity(
    claim_matrix
)


# ================================================================
# 16. CLAIM RELATIONSHIP ENGINE
# ================================================================

relationships = []

for i in range(
    len(claims)
):

    for j in range(
        i + 1,
        len(claims)
    ):

        similarity = float(
            similarity_matrix[i, j]
        )

        if similarity >= SIMILARITY_THRESHOLD:

            same_family = (
                claims.iloc[i]["family"]
                ==
                claims.iloc[j]["family"]
            )

            relationship = (
                "same_family"
                if same_family
                else "semantically_related"
            )

            relationships.append({

                "claim_a":
                    claims.iloc[i]["claim_id"],

                "claim_b":
                    claims.iloc[j]["claim_id"],

                "similarity":
                    round(
                        similarity,
                        4
                    ),

                "relationship":
                    relationship,

                "family_a":
                    claims.iloc[i]["family"],

                "family_b":
                    claims.iloc[j]["family"]
            })


relationships_df = pd.DataFrame(
    relationships
)


if relationships_df.empty:

    relationships_df = pd.DataFrame(
        columns=[
            "claim_a",
            "claim_b",
            "similarity",
            "relationship",
            "family_a",
            "family_b"
        ]
    )


print(
    "\nPotential claim relationships:",
    len(relationships_df)
)


# ================================================================
# 17. TEMPORAL LINEAGE
# ================================================================

claims = claims.sort_values(
    [
        "family",
        "date"
    ]
).reset_index(
    drop=True
)


claims["lineage_position"] = (
    claims
    .groupby("family")
    .cumcount()
    + 1
)


claims["family_size"] = (
    claims
    .groupby("family")["claim_id"]
    .transform("count")
)


# ================================================================
# 18. PREVIOUS CLAIM
# ================================================================

claims["previous_claim_id"] = (
    claims
    .groupby("family")["claim_id"]
    .shift(1)
)


claims["previous_strength"] = (
    claims
    .groupby("family")["claim_strength"]
    .shift(1)
)


claims["previous_evidence"] = (
    claims
    .groupby("family")["evidence_strength"]
    .shift(1)
)


# ================================================================
# 19. CLAIM DRIFT
# ================================================================

claims["claim_drift"] = (
    (
        claims["claim_strength"]
        -
        claims["previous_strength"]
    )
    .abs()
    .fillna(0)
)


# ================================================================
# 20. SEMANTIC DRIFT
# ================================================================

claims["semantic_drift"] = 0.0

for family in claims["family"].unique():

    family_indices = list(
        claims[
            claims["family"] == family
        ].index
    )

    for position in range(
        1,
        len(family_indices)
    ):

        current_idx = (
            family_indices[position]
        )

        previous_idx = (
            family_indices[position - 1]
        )

        similarity = cosine_similarity(

            claim_matrix[
                current_idx:current_idx + 1
            ],

            claim_matrix[
                previous_idx:previous_idx + 1
            ]

        )[0][0]

        claims.loc[
            current_idx,
            "semantic_drift"
        ] = round(
            1 - similarity,
            4
        )


# ================================================================
# 21. CONTEXT RETENTION
# ================================================================

claims["context_retention"] = 1.0


for family in claims["family"].unique():

    family_mask = (
        claims["family"] == family
    )

    family_indices = list(
        claims[family_mask].index
    )

    if not family_indices:
        continue

    first_idx = family_indices[0]

    original_qualifiers = set(
        claims.loc[
            first_idx,
            "qualifiers"
        ]
    )

    for idx in family_indices:

        current_qualifiers = set(
            claims.loc[
                idx,
                "qualifiers"
            ]
        )

        if len(
            original_qualifiers
        ) == 0:

            retention = 1.0

        else:

            retention = (

                len(
                    original_qualifiers
                    &
                    current_qualifiers
                )

                /

                len(
                    original_qualifiers
                )
            )

        claims.loc[
            idx,
            "context_retention"
        ] = round(
            retention,
            4
        )


claims["context_loss"] = (
    1 -
    claims["context_retention"]
)


# ================================================================
# 22. INFORMATION DECAY
# ================================================================

claims["information_decay"] = 0.0


for family in claims["family"].unique():

    family_mask = (
        claims["family"] == family
    )

    family_indices = list(
        claims[family_mask].index
    )

    if not family_indices:
        continue

    first_idx = family_indices[0]

    original_evidence = float(
        claims.loc[
            first_idx,
            "evidence_strength"
        ]
    )

    for idx in family_indices:

        evidence = float(
            claims.loc[
                idx,
                "evidence_strength"
            ]
        )

        evidence_loss = max(
            0,
            original_evidence -
            evidence
        )

        context_loss = float(
            claims.loc[
                idx,
                "context_loss"
            ]
        )

        semantic_loss = float(
            claims.loc[
                idx,
                "semantic_drift"
            ]
        )

        decay = (

            0.45 * evidence_loss

            +

            0.35 * context_loss

            +

            0.20 * semantic_loss
        )

        claims.loc[
            idx,
            "information_decay"
        ] = round(
            min(decay, 1),
            4
        )


# ================================================================
# 23. EVIDENCE–CLAIM DIVERGENCE
# ================================================================

claims["evidence_claim_divergence"] = (

    claims["claim_strength"]

    -

    claims["evidence_strength"]
)


claims["absolute_divergence"] = (
    claims["evidence_claim_divergence"]
    .abs()
)


# ================================================================
# 24. SOURCE DEPENDENCY
# ================================================================

source_count_map = (

    claims
    .groupby("family")["source"]
    .nunique()
    .to_dict()
)


claims["family_source_count"] = (
    claims["family"]
    .map(source_count_map)
)


claims["source_dependency"] = (

    1
    /

    claims["family_source_count"]
)


# ================================================================
# 25. CLAIM MUTATION ENGINE
# ================================================================

def detect_mutations(previous_text, current_text):

    if not isinstance(
        previous_text,
        str
    ):

        return []

    if not isinstance(
        current_text,
        str
    ):

        return []

    previous = previous_text.lower()

    current = current_text.lower()

    mutations = []

    # ------------------------------------------------------------
    # uncertainty → certainty
    # ------------------------------------------------------------

    uncertainty_terms = [
        "may",
        "might",
        "appears",
        "suggests",
        "preliminary"
    ]

    certainty_terms = [
        "proven",
        "causes",
        "is",
        "reduces",
        "improves"
    ]

    previous_uncertain = any(
        term in previous
        for term in uncertainty_terms
    )

    current_certain = any(
        term in current
        for term in certainty_terms
    )

    if (
        previous_uncertain
        and
        current_certain
    ):

        mutations.append(
            "uncertainty_to_certainty"
        )


    # ------------------------------------------------------------
    # association → causation
    # ------------------------------------------------------------

    if (
        "associated" in previous
        and
        "causes" in current
    ):

        mutations.append(
            "association_to_causation"
        )


    # ------------------------------------------------------------
    # preliminary → assertion
    # ------------------------------------------------------------

    if (
        "preliminary" in previous
        and
        "preliminary" not in current
    ):

        mutations.append(
            "preliminary_context_removed"
        )


    # ------------------------------------------------------------
    # dramatic strengthening
    # ------------------------------------------------------------

    if (
        "proven" in current
        and
        "proven" not in previous
    ):

        mutations.append(
            "proven_language_added"
        )


    # ------------------------------------------------------------
    # significant language change
    # ------------------------------------------------------------

    if (
        "significantly" in previous
        and
        "significantly" not in current
    ):

        mutations.append(
            "significance_qualifier_removed"
        )


    return mutations


claims["mutations"] = [[] for _ in range(len(claims))]


for family in claims["family"].unique():

    indices = list(
        claims[
            claims["family"] == family
        ].index
    )

    for pos in range(
        1,
        len(indices)
    ):

        previous_idx = indices[pos - 1]

        current_idx = indices[pos]

        mutations = detect_mutations(

            claims.loc[
                previous_idx,
                "claim"
            ],

            claims.loc[
                current_idx,
                "claim"
            ]
        )

        claims.at[
            current_idx,
            "mutations"
        ] = mutations


# ================================================================
# 26. CONTRADICTION DETECTION
# ================================================================

def is_negative_claim(text):

    t = text.lower()

    return (
        "did not" in t
        or
        "not significantly" in t
        or
        "no significant" in t
    )


def is_positive_claim(text):

    t = text.lower()

    return (
        "reduces" in t
        or
        "reduce" in t
        or
        "improves" in t
        or
        "improve" in t
        or
        "causes" in t
    )


contradiction_records = []


for i in range(
    len(claims)
):

    for j in range(
        i + 1,
        len(claims)
    ):

        similarity = float(
            similarity_matrix[
                claims.index[i],
                claims.index[j]
            ]
        )

        if similarity < 0.25:

            continue

        text_a = claims.iloc[i]["claim"]

        text_b = claims.iloc[j]["claim"]

        opposite = (

            (
                is_negative_claim(text_a)
                and
                is_positive_claim(text_b)
            )

            or

            (
                is_negative_claim(text_b)
                and
                is_positive_claim(text_a)
            )
        )

        if opposite:

            contradiction_records.append({

                "claim_a":
                    claims.iloc[i]["claim_id"],

                "claim_b":
                    claims.iloc[j]["claim_id"],

                "family_a":
                    claims.iloc[i]["family"],

                "family_b":
                    claims.iloc[j]["family"],

                "similarity":
                    round(
                        similarity,
                        4
                    ),

                "claim_a_text":
                    text_a,

                "claim_b_text":
                    text_b
            })


contradictions_df = pd.DataFrame(
    contradiction_records
)


if contradictions_df.empty:

    contradictions_df = pd.DataFrame(
        columns=[
            "claim_a",
            "claim_b",
            "family_a",
            "family_b",
            "similarity",
            "claim_a_text",
            "claim_b_text"
        ]
    )


# ================================================================
# 27. FAMILY ANALYTICS
# ================================================================
#
# IMPORTANT:
#
# Every aggregation uses:
#
#     as_index=False
#
# Therefore "family" remains a regular column.
#
# This directly prevents the KeyError you were getting.
#
# ================================================================

family_stats = (

    claims
    .groupby(
        "family",
        as_index=False
    )
    .agg(

        claim_count=(
            "claim_id",
            "count"
        ),

        source_count=(
            "source",
            "nunique"
        ),

        first_date=(
            "date",
            "min"
        ),

        latest_date=(
            "date",
            "max"
        ),

        average_claim_strength=(
            "claim_strength",
            "mean"
        ),

        average_evidence_strength=(
            "evidence_strength",
            "mean"
        ),

        average_context_retention=(
            "context_retention",
            "mean"
        ),

        average_information_decay=(
            "information_decay",
            "mean"
        ),

        average_divergence=(
            "absolute_divergence",
            "mean"
        ),

        average_claim_drift=(
            "claim_drift",
            "mean"
        ),

        average_semantic_drift=(
            "semantic_drift",
            "mean"
        )
    )
)


# ================================================================
# 28. STRENGTH ANALYTICS
# ================================================================

strength_stats = (

    claims
    .sort_values(
        [
            "family",
            "date"
        ]
    )
    .groupby(
        "family",
        as_index=False
    )
    .agg(

        initial_claim_strength=(
            "claim_strength",
            "first"
        ),

        latest_claim_strength=(
            "claim_strength",
            "last"
        ),

        maximum_claim_strength=(
            "claim_strength",
            "max"
        ),

        minimum_claim_strength=(
            "claim_strength",
            "min"
        )
    )
)


strength_stats["claim_strength_change"] = (

    strength_stats["latest_claim_strength"]

    -

    strength_stats["initial_claim_strength"]
)


# ================================================================
# 29. EVIDENCE ANALYTICS
# ================================================================

evidence_stats = (

    claims
    .sort_values(
        [
            "family",
            "date"
        ]
    )
    .groupby(
        "family",
        as_index=False
    )
    .agg(

        initial_evidence_strength=(
            "evidence_strength",
            "first"
        ),

        latest_evidence_strength=(
            "evidence_strength",
            "last"
        ),

        maximum_evidence_strength=(
            "evidence_strength",
            "max"
        ),

        minimum_evidence_strength=(
            "evidence_strength",
            "min"
        )
    )
)


evidence_stats["evidence_change"] = (

    evidence_stats["latest_evidence_strength"]

    -

    evidence_stats["initial_evidence_strength"]
)


# ================================================================
# 30. CONTEXT ANALYTICS
# ================================================================

context_stats = (

    claims
    .sort_values(
        [
            "family",
            "date"
        ]
    )
    .groupby(
        "family",
        as_index=False
    )
    .agg(

        initial_context_retention=(
            "context_retention",
            "first"
        ),

        latest_context_retention=(
            "context_retention",
            "last"
        ),

        maximum_context_loss=(
            "context_loss",
            "max"
        )
    )
)


# ================================================================
# 31. DECAY ANALYTICS
# ================================================================

decay_stats = (

    claims
    .sort_values(
        [
            "family",
            "date"
        ]
    )
    .groupby(
        "family",
        as_index=False
    )
    .agg(

        initial_information_decay=(
            "information_decay",
            "first"
        ),

        final_information_decay=(
            "information_decay",
            "last"
        ),

        maximum_information_decay=(
            "information_decay",
            "max"
        )
    )
)


# ================================================================
# 32. MUTATION ANALYTICS
# ================================================================

def count_mutations(series):

    return sum(
        len(item)
        for item in series
        if isinstance(item, list)
    )


mutation_stats = (

    claims
    .groupby(
        "family",
        as_index=False
    )
    .agg(

        mutation_events=(
            "mutations",
            count_mutations
        )
    )
)


# ================================================================
# 33. SAFE ANALYTICS MERGE
# ================================================================

analytics_df = family_stats.copy()


# Safety assertion
assert "family" in analytics_df.columns


for extra_df in [

    strength_stats,

    evidence_stats,

    context_stats,

    decay_stats,

    mutation_stats

]:

    assert "family" in extra_df.columns

    analytics_df = analytics_df.merge(

        extra_df,

        on="family",

        how="left",

        validate="one_to_one"
    )


# ================================================================
# 34. RESEARCH INTEGRITY SIGNAL
# ================================================================

def calculate_integrity_signal(row):

    divergence = min(
        abs(
            float(
                row["average_divergence"]
            )
        ),
        1
    )

    decay = min(
        max(
            float(
                row["final_information_decay"]
            ),
            0
        ),
        1
    )

    context_loss = min(
        max(
            1 -
            float(
                row["latest_context_retention"]
            ),
            0
        ),
        1
    )

    semantic_drift = min(
        max(
            float(
                row["average_semantic_drift"]
            ),
            0
        ),
        1
    )

    source_dependency = (

        1 /

        max(
            float(
                row["source_count"]
            ),
            1
        )
    )

    score = (

        0.25 * divergence

        +

        0.25 * decay

        +

        0.20 * context_loss

        +

        0.15 * semantic_drift

        +

        0.15 * source_dependency
    )

    return round(
        score,
        4
    )


analytics_df[
    "research_integrity_signal"
] = analytics_df.apply(
    calculate_integrity_signal,
    axis=1
)


# ================================================================
# 35. FAMILY STATUS
# ================================================================

def family_status(row):

    if row["research_integrity_signal"] >= 0.35:

        return "High analytical signal"

    if row["research_integrity_signal"] >= 0.20:

        return "Moderate analytical signal"

    return "Low analytical signal"


analytics_df["signal_band"] = (
    analytics_df.apply(
        family_status,
        axis=1
    )
)


# ================================================================
# 36. CLAIM SEARCH ENGINE
# ================================================================

def search_claims(
    query,
    top_k=5
):

    query_vector = (
        vectorizer.transform(
            [
                normalize_text(
                    query
                )
            ]
        )
    )

    scores = (
        cosine_similarity(
            query_vector,
            claim_matrix
        )[0]
    )

    result = claims.copy()

    result["search_score"] = scores

    result = result.sort_values(
        "search_score",
        ascending=False
    )

    columns = [

        "claim_id",

        "date",

        "source",

        "family",

        "claim",

        "modality",

        "claim_strength",

        "evidence_strength",

        "context_retention",

        "information_decay",

        "search_score"
    ]

    return (
        result[
            columns
        ]
        .head(top_k)
        .reset_index(drop=True)
    )


# ================================================================
# 37. CLAIM DNA SEARCH
# ================================================================

def get_claim_dna(
    claim_id
):

    matches = claims[
        claims["claim_id"].str.upper()
        ==
        str(claim_id).upper()
    ]

    if matches.empty:

        return {
            "error":
                f"Claim {claim_id} not found."
        }

    row = matches.iloc[0]

    return {

        "claim_id":
            row["claim_id"],

        "claim":
            row["claim"],

        "date":
            str(
                row["date"].date()
            ),

        "source":
            row["source"],

        "source_type":
            row["source_type"],

        "family":
            row["family"],

        "subject":
            row["subject"],

        "predicate":
            row["predicate"],

        "object":
            row["object"],

        "modality":
            row["modality"],

        "qualifiers":
            row["qualifiers"],

        "entities":
            row["entities"],

        "claim_strength":
            float(
                row["claim_strength"]
            ),

        "evidence_strength":
            float(
                row["evidence_strength"]
            ),

        "context_retention":
            float(
                row["context_retention"]
            ),

        "information_decay":
            float(
                row["information_decay"]
            ),

        "mutations":
            row["mutations"]
    }


# ================================================================
# 38. LINEAGE ENGINE
# ================================================================

def get_lineage(
    claim_id
):

    matches = claims[
        claims["claim_id"].str.upper()
        ==
        str(claim_id).upper()
    ]

    if matches.empty:

        return pd.DataFrame()

    family = matches.iloc[0][
        "family"
    ]

    lineage = (
        claims[
            claims["family"] == family
        ]
        .sort_values("date")
        .reset_index(drop=True)
    )

    columns = [

        "claim_id",

        "date",

        "source",

        "source_type",

        "claim",

        "modality",

        "claim_strength",

        "evidence_strength",

        "claim_drift",

        "semantic_drift",

        "context_retention",

        "information_decay",

        "evidence_claim_divergence",

        "mutations"
    ]

    return lineage[
        columns
    ]


# ================================================================
# 39. RESEARCH COPILOT
# ================================================================

def research_copilot(
    query
):

    results = search_claims(
        query,
        top_k=5
    )

    if results.empty:

        return (
            "No relevant claims found."
        )

    best = results.iloc[0]

    family = best["family"]

    lineage = (
        claims[
            claims["family"] == family
        ]
        .sort_values("date")
        .reset_index(drop=True)
    )

    first = lineage.iloc[0]

    latest = lineage.iloc[-1]

    strength_change = (

        float(
            latest["claim_strength"]
        )

        -

        float(
            first["claim_strength"]
        )
    )

    evidence_change = (

        float(
            latest["evidence_strength"]
        )

        -

        float(
            first["evidence_strength"]
        )
    )

    output = []

    output.append(
        "\nTRUTHMESH RESEARCH COPILOT"
    )

    output.append(
        "=" * 70
    )

    output.append(
        f"Query: {query}"
    )

    output.append(
        f"Matched family: {family}"
    )

    output.append(
        f"Versions tracked: {len(lineage)}"
    )

    output.append(
        f"Sources tracked: {lineage['source'].nunique()}"
    )

    output.append(
        "\nORIGINAL CLAIM"
    )

    output.append(
        str(first["claim"])
    )

    output.append(
        "\nLATEST CLAIM"
    )

    output.append(
        str(latest["claim"])
    )

    output.append(
        "\nNUMERICAL EVOLUTION"
    )

    output.append(
        f"Claim strength change: "
        f"{strength_change:+.4f}"
    )

    output.append(
        f"Evidence change: "
        f"{evidence_change:+.4f}"
    )

    output.append(
        f"Context retention: "
        f"{latest['context_retention']:.4f}"
    )

    output.append(
        f"Information decay: "
        f"{latest['information_decay']:.4f}"
    )

    output.append(
        f"Evidence-claim divergence: "
        f"{latest['evidence_claim_divergence']:+.4f}"
    )

    all_mutations = []

    for item in lineage[
        "mutations"
    ]:

        if isinstance(
            item,
            list
        ):

            all_mutations.extend(
                item
            )

    output.append(
        "\nCLAIM MUTATIONS"
    )

    if all_mutations:

        for mutation in dict.fromkeys(
            all_mutations
        ):

            output.append(
                f"• {mutation}"
            )

    else:

        output.append(
            "• No detected mutation event"
        )

    return "\n".join(
        output
    )


# ================================================================
# 40. KNOWLEDGE GRAPH
# ================================================================

G = nx.DiGraph()


# ---------------------------------------------------------------
# CLAIM NODES
# ---------------------------------------------------------------

for _, row in claims.iterrows():

    claim_node = (
        "CLAIM:"
        +
        row["claim_id"]
    )

    G.add_node(

        claim_node,

        node_type="claim",

        label=row["claim_id"],

        family=row["family"],

        strength=float(
            row["claim_strength"]
        )
    )


# ---------------------------------------------------------------
# SOURCE NODES
# ---------------------------------------------------------------

for source in claims[
    "source"
].unique():

    source_node = (
        "SOURCE:"
        +
        source
    )

    G.add_node(

        source_node,

        node_type="source",

        label=source
    )


# ---------------------------------------------------------------
# FAMILY NODES
# ---------------------------------------------------------------

for family in claims[
    "family"
].unique():

    family_node = (
        "FAMILY:"
        +
        family
    )

    G.add_node(

        family_node,

        node_type="family",

        label=family
    )


# ---------------------------------------------------------------
# ENTITY NODES
# ---------------------------------------------------------------

all_entities = []

for entities in claims[
    "entities"
]:

    all_entities.extend(
        entities
    )


for entity in set(
    all_entities
):

    entity_node = (
        "ENTITY:"
        +
        entity
    )

    G.add_node(

        entity_node,

        node_type="entity",

        label=entity
    )


# ---------------------------------------------------------------
# EDGES
# ---------------------------------------------------------------

for _, row in claims.iterrows():

    claim_node = (
        "CLAIM:"
        +
        row["claim_id"]
    )

    source_node = (
        "SOURCE:"
        +
        row["source"]
    )

    family_node = (
        "FAMILY:"
        +
        row["family"]
    )

    G.add_edge(

        claim_node,

        source_node,

        relation="supported_by"
    )

    G.add_edge(

        claim_node,

        family_node,

        relation="belongs_to"
    )

    for entity in row[
        "entities"
    ]:

        entity_node = (
            "ENTITY:"
            +
            entity
        )

        G.add_edge(

            claim_node,

            entity_node,

            relation="mentions"
        )


# ---------------------------------------------------------------
# TEMPORAL LINEAGE EDGES
# ---------------------------------------------------------------

for family in claims[
    "family"
].unique():

    family_data = (
        claims[
            claims["family"] == family
        ]
        .sort_values("date")
    )

    ids = list(
        family_data["claim_id"]
    )

    for i in range(
        1,
        len(ids)
    ):

        previous_node = (
            "CLAIM:"
            +
            ids[i - 1]
        )

        current_node = (
            "CLAIM:"
            +
            ids[i]
        )

        G.add_edge(

            previous_node,

            current_node,

            relation="temporal_evolution"
        )


# ---------------------------------------------------------------
# SEMANTIC RELATIONSHIP EDGES
# ---------------------------------------------------------------

for _, row in relationships_df.iterrows():

    a = (
        "CLAIM:"
        +
        row["claim_a"]
    )

    b = (
        "CLAIM:"
        +
        row["claim_b"]
    )

    G.add_edge(

        a,

        b,

        relation=row["relationship"],

        similarity=float(
            row["similarity"]
        )
    )


# ================================================================
# 41. GRAPH STATISTICS
# ================================================================

graph_statistics = {

    "total_nodes":
        G.number_of_nodes(),

    "total_edges":
        G.number_of_edges(),

    "claim_nodes":
        sum(
            1
            for _, data
            in G.nodes(data=True)
            if data.get(
                "node_type"
            ) == "claim"
        ),

    "source_nodes":
        sum(
            1
            for _, data
            in G.nodes(data=True)
            if data.get(
                "node_type"
            ) == "source"
        ),

    "family_nodes":
        sum(
            1
            for _, data
            in G.nodes(data=True)
            if data.get(
                "node_type"
            ) == "family"
        ),

    "entity_nodes":
        sum(
            1
            for _, data
            in G.nodes(data=True)
            if data.get(
                "node_type"
            ) == "entity"
        )
}


# ================================================================
# 42. CLAIM TIME MACHINE
# ================================================================

def claim_time_machine(
    family_name
):

    timeline = (

        claims[
            claims["family"]
            ==
            family_name
        ]

        .sort_values("date")
    )

    if timeline.empty:

        print(
            "Family not found."
        )

        return None


    fig = go.Figure()


    fig.add_trace(

        go.Scatter(

            x=timeline["date"],

            y=timeline[
                "claim_strength"
            ],

            mode=
                "lines+markers+text",

            text=timeline[
                "claim_id"
            ],

            textposition=
                "top center",

            customdata=
                timeline[
                    [
                        "claim",
                        "source",
                        "evidence_strength"
                    ]
                ].values,

            hovertemplate=

                "<b>%{text}</b><br>"
                +

                "Date: %{x}<br>"
                +

                "Claim Strength: %{y:.3f}<br>"
                +

                "Source: %{customdata[1]}<br>"
                +

                "Evidence: %{customdata[2]:.3f}<br>"
                +

                "%{customdata[0]}"
                +

                "<extra></extra>"
        )
    )


    fig.update_layout(

        title=
            f"Claim Time Machine — {family_name}",

        xaxis_title=
            "Timeline",

        yaxis_title=
            "Claim Strength",

        template=
            "plotly_dark",

        height=600
    )


    fig.show()

    return timeline


# ================================================================
# 43. DASHBOARD 1 — CLAIM STRENGTH
# ================================================================

fig_strength = px.line(

    claims.sort_values("date"),

    x="date",

    y="claim_strength",

    color="family",

    markers=True,

    hover_data=[

        "claim_id",

        "source",

        "claim",

        "evidence_strength",

        "modality"
    ],

    title=
        "TRUTHMESH — Claim Strength Evolution"
)


fig_strength.update_layout(

    template="plotly_dark",

    height=650
)


fig_strength.show()


# ================================================================
# 44. DASHBOARD 2 — EVIDENCE VS CLAIM
# ================================================================

fig_evidence = px.scatter(

    claims,

    x="evidence_strength",

    y="claim_strength",

    color="family",

    size="sample_size",

    hover_data=[

        "claim_id",

        "source",

        "claim",

        "modality"
    ],

    title=
        "TRUTHMESH — Evidence vs Claim Strength"
)


fig_evidence.add_shape(

    type="line",

    x0=0,

    y0=0,

    x1=1,

    y1=1,

    line=dict(
        dash="dash"
    )
)


fig_evidence.update_layout(

    template="plotly_dark",

    height=650
)


fig_evidence.show()


# ================================================================
# 45. DASHBOARD 3 — CONTEXT RETENTION
# ================================================================

fig_context = px.line(

    claims.sort_values("date"),

    x="date",

    y="context_retention",

    color="family",

    markers=True,

    hover_data=[

        "claim_id",

        "source",

        "claim",

        "qualifiers"
    ],

    title=
        "TRUTHMESH — Context Retention"
)


fig_context.update_layout(

    template="plotly_dark",

    height=650
)


fig_context.show()


# ================================================================
# 46. DASHBOARD 4 — INFORMATION DECAY
# ================================================================

fig_decay = px.line(

    claims.sort_values("date"),

    x="date",

    y="information_decay",

    color="family",

    markers=True,

    hover_data=[

        "claim_id",

        "source",

        "claim"
    ],

    title=
        "TRUTHMESH — Information Decay"
)


fig_decay.update_layout(

    template="plotly_dark",

    height=650
)


fig_decay.show()


# ================================================================
# 47. DASHBOARD 5 — RESEARCH SIGNAL
# ================================================================

fig_signal = px.bar(

    analytics_df,

    x="family",

    y="research_integrity_signal",

    color="signal_band",

    hover_data=[

        "claim_count",

        "source_count",

        "average_divergence",

        "final_information_decay",

        "latest_context_retention",

        "mutation_events"
    ],

    title=
        "TRUTHMESH — Family-Level Analytical Signal"
)


fig_signal.update_layout(

    template="plotly_dark",

    height=650,

    xaxis_tickangle=-30
)


fig_signal.show()


# ================================================================
# 48. DASHBOARD 6 — CLAIM DRIFT
# ================================================================

fig_drift = px.scatter(

    claims,

    x="date",

    y="claim_drift",

    color="family",

    size="semantic_drift",

    hover_data=[

        "claim_id",

        "claim",

        "modality",

        "mutations"
    ],

    title=
        "TRUTHMESH — Claim Drift & Mutation"
)


fig_drift.update_layout(

    template="plotly_dark",

    height=650
)


fig_drift.show()


# ================================================================
# 49. KNOWLEDGE GRAPH VISUALIZATION
# ================================================================

def visualize_knowledge_graph():

    # Use a subset for readability
    graph = G.copy()

    positions = nx.spring_layout(

        graph,

        seed=RANDOM_STATE,

        k=1.4,

        iterations=80
    )


    edge_x = []

    edge_y = []


    for source, target in graph.edges():

        x0, y0 = positions[
            source
        ]

        x1, y1 = positions[
            target
        ]

        edge_x.extend(
            [
                x0,
                x1,
                None
            ]
        )

        edge_y.extend(
            [
                y0,
                y1,
                None
            ]
        )


    edge_trace = go.Scatter(

        x=edge_x,

        y=edge_y,

        mode="lines",

        line=dict(
            width=0.7
        ),

        hoverinfo="none"
    )


    node_x = []

    node_y = []

    node_text = []

    node_sizes = []


    for node, data in graph.nodes(
        data=True
    ):

        x, y = positions[
            node
        ]

        node_x.append(x)

        node_y.append(y)

        node_text.append(
            data.get(
                "label",
                node
            )
        )


        node_type = data.get(
            "node_type"
        )


        if node_type == "claim":

            node_sizes.append(18)

        elif node_type == "source":

            node_sizes.append(13)

        elif node_type == "family":

            node_sizes.append(24)

        else:

            node_sizes.append(15)


    node_trace = go.Scatter(

        x=node_x,

        y=node_y,

        mode=
            "markers+text",

        text=node_text,

        textposition=
            "top center",

        marker=dict(
            size=node_sizes
        ),

        hoverinfo="text"
    )


    fig = go.Figure(

        data=[

            edge_trace,

            node_trace
        ]
    )


    fig.update_layout(

        title=
            "TRUTHMESH Knowledge Graph",

        template=
            "plotly_dark",

        showlegend=False,

        height=800,

        xaxis=dict(

            showgrid=False,

            zeroline=False,

            showticklabels=False
        ),

        yaxis=dict(

            showgrid=False,

            zeroline=False,

            showticklabels=False
        )
    )


    fig.show()


visualize_knowledge_graph()


# ================================================================
# 50. FAMILY ANALYTICS TABLE
# ================================================================

print("\n")
print("=" * 90)
print("                    FAMILY ANALYTICS")
print("=" * 90)


family_display_columns = [

    "family",

    "claim_count",

    "source_count",

    "average_claim_strength",

    "average_evidence_strength",

    "latest_claim_strength",

    "latest_evidence_strength",

    "claim_strength_change",

    "latest_context_retention",

    "final_information_decay",

    "average_divergence",

    "mutation_events",

    "research_integrity_signal"
]


display(

    analytics_df[
        family_display_columns
    ]
    .round(4)
)


# ================================================================
# 51. CLAIM DNA EXAMPLE
# ================================================================

print("\n")
print("=" * 90)
print("                         CLAIM DNA")
print("=" * 90)


for claim_id in [

    "C001",

    "C004",

    "C005",

    "C008",

    "C015"

]:

    dna = get_claim_dna(
        claim_id
    )

    print("\n")
    print(
        f"CLAIM {claim_id}"
    )

    print("-" * 70)

    for key, value in dna.items():

        print(
            f"{key:25}: {value}"
        )


# ================================================================
# 52. CLAIM SEARCH DEMO
# ================================================================

print("\n")
print("=" * 90)
print("                      CLAIM SEARCH")
print("=" * 90)


search_result = search_claims(

    "gene disease association",

    top_k=5
)


display(
    search_result.round(4)
)


# ================================================================
# 53. RESEARCH COPILOT DEMO
# ================================================================

print(
    research_copilot(
        "Gene X disease"
    )
)


# ================================================================
# 54. CLAIM TIME MACHINE DEMO
# ================================================================

print("\n")
print("=" * 90)
print("                   CLAIM TIME MACHINE")
print("=" * 90)


claim_time_machine(
    "Gene X – Disease Y"
)


# ================================================================
# 55. CONTRADICTION REPORT
# ================================================================

print("\n")
print("=" * 90)
print("                  CONTRADICTION REPORT")
print("=" * 90)


if contradictions_df.empty:

    print(
        "No potential contradictions detected."
    )

else:

    display(
        contradictions_df
    )


# ================================================================
# 56. KNOWLEDGE GRAPH REPORT
# ================================================================

print("\n")
print("=" * 90)
print("                KNOWLEDGE GRAPH REPORT")
print("=" * 90)


for key, value in graph_statistics.items():

    print(
        f"{key:25}: {value}"
    )


# ================================================================
# 57. MUTATION REPORT
# ================================================================

print("\n")
print("=" * 90)
print("                    CLAIM MUTATIONS")
print("=" * 90)


mutation_rows = []


for _, row in claims.iterrows():

    if isinstance(
        row["mutations"],
        list
    ):

        for mutation in row[
            "mutations"
        ]:

            mutation_rows.append({

                "claim_id":
                    row["claim_id"],

                "family":
                    row["family"],

                "date":
                    row["date"],

                "mutation":
                    mutation,

                "claim":
                    row["claim"]
            })


mutation_df = pd.DataFrame(
    mutation_rows
)


if mutation_df.empty:

    print(
        "No mutation events detected."
    )

else:

    display(
        mutation_df
    )


# ================================================================
# 58. EXPORT CLEAN CLAIM DATA
# ================================================================

export_claims = claims.copy()


# Convert lists/dictionaries to JSON strings
export_claims[
    "entities"
] = export_claims[
    "entities"
].apply(
    lambda x: json.dumps(
        x,
        ensure_ascii=False
    )
)


export_claims[
    "qualifiers"
] = export_claims[
    "qualifiers"
].apply(
    lambda x: json.dumps(
        x,
        ensure_ascii=False
    )
)


export_claims[
    "mutations"
] = export_claims[
    "mutations"
].apply(
    lambda x: json.dumps(
        x,
        ensure_ascii=False
    )
)


export_claims[
    "claim_dna"
] = export_claims[
    "claim_dna"
].apply(
    lambda x: json.dumps(
        x,
        ensure_ascii=False
    )
)


export_claims.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "truthmesh_claims.csv"
    ),

    index=False
)


# ================================================================
# 59. EXPORT FAMILY ANALYTICS
# ================================================================

analytics_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "truthmesh_family_analytics.csv"
    ),

    index=False
)


# ================================================================
# 60. EXPORT RELATIONSHIPS
# ================================================================

relationships_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "truthmesh_relationships.csv"
    ),

    index=False
)


# ================================================================
# 61. EXPORT CONTRADICTIONS
# ================================================================

contradictions_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "truthmesh_contradictions.csv"
    ),

    index=False
)


# ================================================================
# 62. EXPORT MUTATIONS
# ================================================================

mutation_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "truthmesh_mutations.csv"
    ),

    index=False
)


# ================================================================
# 63. EXPORT GRAPH EDGES
# ================================================================

graph_edges = []

for source, target, data in G.edges(
    data=True
):

    graph_edges.append({

        "source": source,

        "target": target,

        "relation":
            data.get(
                "relation",
                ""
            ),

        "similarity":
            data.get(
                "similarity",
                np.nan
            )
    })


graph_edges_df = pd.DataFrame(
    graph_edges
)


graph_edges_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "truthmesh_graph_edges.csv"
    ),

    index=False
)


# ================================================================
# 64. EXPORT JSON REPORT
# ================================================================

project_summary = {

    "project":
        PROJECT_NAME,

    "version":
        VERSION,

    "total_claims":
        int(
            len(claims)
        ),

    "unique_sources":
        int(
            claims[
                "source"
            ].nunique()
        ),

    "claim_families":
        int(
            claims[
                "family"
            ].nunique()
        ),

    "claim_relationships":
        int(
            len(
                relationships_df
            )
        ),

    "potential_contradictions":
        int(
            len(
                contradictions_df
            )
        ),

    "mutation_events":
        int(
            len(
                mutation_df
            )
        ),

    "graph_nodes":
        int(
            G.number_of_nodes()
        ),

    "graph_edges":
        int(
            G.number_of_edges()
        ),

    "output_directory":
        OUTPUT_DIR
}


with open(

    os.path.join(
        OUTPUT_DIR,
        "truthmesh_project_report.json"
    ),

    "w",

    encoding="utf-8"

) as f:

    json.dump(

        project_summary,

        f,

        indent=4,

        ensure_ascii=False
    )


# ================================================================
# 65. FINAL REPORT
# ================================================================

print("\n")
print("=" * 90)
print("                 TRUTHMESH AI v2.0")
print("                  FINAL REPORT")
print("=" * 90)


print(
    f"\nTotal claims              : "
    f"{len(claims)}"
)


print(
    f"Unique sources            : "
    f"{claims['source'].nunique()}"
)


print(
    f"Claim families            : "
    f"{claims['family'].nunique()}"
)


print(
    f"Claim relationships       : "
    f"{len(relationships_df)}"
)


print(
    f"Potential contradictions  : "
    f"{len(contradictions_df)}"
)


print(
    f"Mutation events           : "
    f"{len(mutation_df)}"
)


print(
    f"Knowledge graph nodes     : "
    f"{G.number_of_nodes()}"
)


print(
    f"Knowledge graph edges     : "
    f"{G.number_of_edges()}"
)


# ================================================================
# 66. FAMILY INTELLIGENCE REPORT
# ================================================================

print("\n")
print(
    "=" * 90
)

print(
    "                 FAMILY INTELLIGENCE"
)

print(
    "=" * 90
)


for _, row in analytics_df.iterrows():

    print("\n")

    print(
        f"FAMILY: {row['family']}"
    )

    print(
        "-" * 70
    )

    print(
        f"Claim versions          : "
        f"{int(row['claim_count'])}"
    )

    print(
        f"Sources                 : "
        f"{int(row['source_count'])}"
    )

    print(
        f"Initial claim strength  : "
        f"{row['initial_claim_strength']:.4f}"
    )

    print(
        f"Latest claim strength   : "
        f"{row['latest_claim_strength']:.4f}"
    )

    print(
        f"Claim strength change   : "
        f"{row['claim_strength_change']:+.4f}"
    )

    print(
        f"Initial evidence        : "
        f"{row['initial_evidence_strength']:.4f}"
    )

    print(
        f"Latest evidence         : "
        f"{row['latest_evidence_strength']:.4f}"
    )

    print(
        f"Evidence change         : "
        f"{row['evidence_change']:+.4f}"
    )

    print(
        f"Context retention       : "
        f"{row['latest_context_retention']:.4f}"
    )

    print(
        f"Information decay       : "
        f"{row['final_information_decay']:.4f}"
    )

    print(
        f"Average divergence      : "
        f"{row['average_divergence']:.4f}"
    )

    print(
        f"Mutation events         : "
        f"{int(row['mutation_events'])}"
    )

    print(
        f"Analytical signal       : "
        f"{row['research_integrity_signal']:.4f}"
    )


# ================================================================
# 67. OUTPUT FILE REPORT
# ================================================================

print("\n")
print(
    "=" * 90
)

print(
    "                   EXPORTED FILES"
)

print(
    "=" * 90
)


for filename in sorted(
    os.listdir(
        OUTPUT_DIR
    )
):

    print(
        f"✓ {OUTPUT_DIR}/{filename}"
    )


# ================================================================
# 68. FINAL API-LIKE OBJECT
# ================================================================

TRUTHMESH = {

    "project":
        PROJECT_NAME,

    "version":
        VERSION,

    "data":
        claims,

    "family_analytics":
        analytics_df,

    "relationships":
        relationships_df,

    "contradictions":
        contradictions_df,

    "mutations":
        mutation_df,

    "knowledge_graph":
        G,

    "vectorizer":
        vectorizer,

    "claim_matrix":
        claim_matrix,

    "functions": {

        "search_claims":
            search_claims,

        "get_claim_dna":
            get_claim_dna,

        "get_lineage":
            get_lineage,

        "research_copilot":
            research_copilot,

        "claim_time_machine":
            claim_time_machine
    },

    "summary":
        project_summary
}


# ================================================================
# 69. SUCCESS
# ================================================================

print("\n")
print("=" * 90)
print("              ✓ TRUTHMESH AI v2.0 IS READY")
print("=" * 90)

print(
    "\nYou can now use:"
)

print(
    "  TRUTHMESH['data']"
)

print(
    "  TRUTHMESH['family_analytics']"
)

print(
    "  TRUTHMESH['relationships']"
)

print(
    "  TRUTHMESH['contradictions']"
)

print(
    "  TRUTHMESH['mutations']"
)

print(
    "  TRUTHMESH['knowledge_graph']"
)

print(
    "  TRUTHMESH['functions']['search_claims'](...)"
)

print(
    "  TRUTHMESH['functions']['get_claim_dna'](...)"
)

print(
    "  TRUTHMESH['functions']['get_lineage'](...)"
)

print(
    "  TRUTHMESH['functions']['research_copilot'](...)"
)

print("\n")
print(
    f"All files saved in: {OUTPUT_DIR}/"
)

print("\n")
print(
    "TRUTHMESH AI PIPELINE COMPLETED SUCCESSFULLY."
)

print("=" * 90)
```
